```python
# Once we get delta credentials we can use this template code
import os
from dotenv import load_dotenv

import ccxt

api_key = os.getenv('API_KEY')
secret = os.getenv('SECRET')

delta_exchange = ccxt.delta({
    'apiKey': api_key,
    'secret': secret,
})
```

```python
markets = delta_exchange.load_markets()
options_markets = [market for market in markets.values() if market['type'] == 'option']
print(options_markets)
```

```python
for option in options_markets:
    print(option)
```
